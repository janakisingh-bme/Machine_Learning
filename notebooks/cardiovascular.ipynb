```python
clean_df["MAP_mmHg"] = (
    clean_df["Diastolic_BP_mmHg"] +
    (clean_df["Systolic_BP_mmHg"] -
     clean_df["Diastolic_BP_mmHg"]) / 3
)
```

```python
clean_df[
    [
        "Systolic_BP_mmHg",
        "Diastolic_BP_mmHg",
        "MAP_mmHg"
    ]
].head()
```

```python
clean_df["Pulse_Pressure_mmHg"] = (
    clean_df["Systolic_BP_mmHg"] -
    clean_df["Diastolic_BP_mmHg"]
)
```

```python
clean_df[
    [
        "Systolic_BP_mmHg",
        "Diastolic_BP_mmHg",
        "Pulse_Pressure_mmHg"
    ]
].head()
```

```python
clean_df["Absolute_ST_Deviation_mV"] = (
    clean_df["ST_Deviation_mV"].abs()
)
```

```python
clean_df[
    [
        "ST_Deviation_mV",
        "Absolute_ST_Deviation_mV"
    ]
].head()
```

```python
clean_df["QRS_Prolonged"] = (
    clean_df["QRS_Duration_ms"] > 120
).astype(int)
```

```python
clean_df["QRS_Prolonged"].value_counts()
```

```python
clean_df["Heart_Rate_Category"] = pd.cut(
    clean_df["Heart_Rate_bpm"],
    bins=[0, 60, 100, np.inf],
    labels=["Bradycardia", "Normal", "Tachycardia"]
)
```

```python
clean_df["Heart_Rate_Category"].value_counts()
```

```python
clean_df[
    [
        "Heart_Rate_bpm",
        "Heart_Rate_Category"
    ]
].head(10)
```

```python
clean_df["Glucose_Category"] = pd.cut(
    clean_df["Glucose_mg_dL"],
    bins=[0, 100, 126, np.inf],
    labels=["Normal", "Elevated", "High"]
)
```

```python
clean_df["Glucose_Category"].value_counts()
```

```python
clean_df["Age_Group"] = pd.cut(
    clean_df["Age"],
    bins=[17, 30, 45, 60, 75, 100],
    labels=[
        "18-30",
        "31-45",
        "46-60",
        "61-75",
        "76+"
    ]
)
```

```python
clean_df["Age_Group"].value_counts().sort_index()
```

```python
clean_df["QRS_QT_Ratio"] = (
    clean_df["QRS_Duration_ms"] /
    clean_df["QT_Interval_ms"]
)
```

```python
clean_df[
    [
        "QRS_Duration_ms",
        "QT_Interval_ms",
        "QRS_QT_Ratio"
    ]
].head()
```

```python
clean_df["Admission_Date"] = pd.to_datetime(
    clean_df["Admission_Date"],
    errors="coerce"
)
```

```python
clean_df[
    [
        "Admission_Date",
        "Admission_Year",
        "Admission_Month",
        "Admission_DayOfWeek"
    ]
].head()
```
